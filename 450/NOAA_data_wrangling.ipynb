### Damage Values

DAMAGE_PROPERTY and DAMAGE_CROPS are currently stored as text. They need to be converted to numeric dollar amounts. 

```python
df_clean["DAMAGE_PROPERTY"].dropna().str[-1].value_counts()
```

```python
df_clean["DAMAGE_CROPS"].dropna().str[-1].value_counts()
```

```python
def convert_damage(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value.endswith("K"):
        return float(value[:-1]) * 1_000
    elif value.endswith("M"):
        return float(value[:-1]) * 1_000_000
    elif value.endswith("B"):
        return float(value[:-1]) * 1_000_000_000
    else:
        return float(value)
```

```python
df_clean["DAMAGE_PROPERTY"] = df_clean["DAMAGE_PROPERTY"].apply(convert_damage)
df_clean["DAMAGE_CROPS"] = df_clean["DAMAGE_CROPS"].apply(convert_damage)
```

```python
df_clean = df_clean.rename(columns={
    "DAMAGE_PROPERTY": "PROPERTY_DAMAGE_USD",
    "DAMAGE_CROPS": "CROP_DAMAGE_USD"
})
df_clean[["PROPERTY_DAMAGE_USD", "CROP_DAMAGE_USD"]].dtypes
```

## 10. Feature Engineering

New variables will be created from the cleaned dataset based on the project's research questions and modeling requirements. 

### Storm Duration

Caluculated using beginning and end date/time of each event. 

Can be used as a potential predictor.

```python
df_clean['STORM_DURATION_HOURS'] = (
    df_clean['END_DATE_TIME'] - df_clean['BEGIN_DATE_TIME']).dt.total_seconds() / 3600
```

```python
df_clean["STORM_DURATION_HOURS"].describe()
```

### Season

```python
df_clean['SEASON'] = df_clean['BEGIN_DATE_TIME'].dt.month.map({
    12: "Winter",
    1: "Winter",
    2: "Winter",
    3: "Spring",
    4: "Spring",
    5: "Spring",
    6: "Summer",
    7: "Summer",
    8: "Summer",
    9: "Fall",
    10: "Fall",
    11: "Fall"
})
    
```

### Human Impact

Features will be created for TOTAL_INJURIES and TOTAL_FATALITIES using the indirect and direct human impact columns. There will also be a feature created for ANY_HUMAN_IMPACT, do determine whether there was human impact from the event. 

```python
df_clean['TOTAL_INJURIES'] = (
    df_clean['INJURIES_DIRECT'] +
    df_clean['INJURIES_INDIRECT'] 
)

df_clean['TOTAL_FATALITIES'] = (
    df_clean['DEATHS_DIRECT'] +
    df_clean['DEATHS_INDIRECT']
)

df_clean['ANY_HUMAN_IMPACT'] = (
    (df_clean['TOTAL_INJURIES'] + df_clean['TOTAL_FATALITIES']) > 0
).astype(int)
```

```python
df_clean[
    [
        "INJURIES_DIRECT",
        "INJURIES_INDIRECT",
        "TOTAL_INJURIES",
        "DEATHS_DIRECT",
        "DEATHS_INDIRECT",
        "TOTAL_FATALITIES",
        "ANY_HUMAN_IMPACT"
    ]
].head(5)
```

### Economic Impact

Property damage and crop damage features have already been converted from string to floats. They are now PROPERTY_DAMAGE_USD and CROP_DAMAGE_USD. A new feature will now be created that describes the TOTAL_ECONOMIC_IMPACT, which will add the values of the two damage columns together. 

A feature will also be created that gives a value of 1 to any total damage that is over $100,000 and is considered MAJOR_PROPERTY_DAMAGE.

```python
df_clean['TOTAL_ECONOMIC_IMPACT_USD'] = (
    df_clean['PROPERTY_DAMAGE_USD'] +
    df_clean['CROP_DAMAGE_USD']
)
```

```python
df_clean['MAJOR_PROPERTY_DAMAGE'] = (
    df_clean['TOTAL_ECONOMIC_IMPACT_USD'] > 100000
).astype(int)

df_clean['MAJOR_PROPERTY_DAMAGE'].value_counts()
```
